1. What is NumPy, and why is it widely used in Python?
   - NumPy (Numerical Python) is a popular open-source Python library used for numerical computing, mathematical operations, scientific computing, and data analysis. It provides a powerful data structure called the NumPy array (ndarray), which can store large amounts of numerical data efficiently.

     Python lists can also store numbers, but NumPy arrays are specifically designed for numerical calculations. NumPy provides functions for mathematical operations, statistics, linear algebra, random number generation, array manipulation, and many other scientific tasks.

NumPy is widely used:
- Fast numerical calculations: NumPy performs calculations much faster than normal Python lists for large numerical datasets.

- Multidimensional arrays: It supports one-dimensional, two-dimensional, and higher-dimensional arrays.

- Memory efficient: NumPy arrays generally consume less memory than Python lists for numerical data.

- Vectorized operations: Mathematical operations can be performed on entire arrays without writing explicit loops.

- Mathematical functions: It provides functions for statistics, trigonometry, logarithms, exponentials, etc.

-  Foundation for other libraries: Libraries such as Pandas, SciPy, Scikit-learn, and many scientific tools use NumPy internally or work closely with it.

2. How does broadcasting work in NumPy?    
   - Broadcasting is a feature of NumPy that allows arithmetic operations to be performed between arrays of different shapes without manually reshaping or copying the smaller array.

     Normally, mathematical operations require arrays to have compatible dimensions. Broadcasting automatically expands the smaller array logically so that the operation can be performed.

 4. Explain the use of the groupby() method in Pandas?
     - The groupby() method in Pandas is used to group rows in a DataFrame based on one or more columns so you can perform aggregate calculations, transformations, or filtering on those groups. It follows the Split-Apply-Combine strategy.

  - Split: The data is broken into chunks based on the keys you specify.  
  -   Apply: A function (like sum, average, or count) is computed for each group independently.
  - Combine: The results of those calculations are merged back into a new DataFrame or Series.

8.  What does the term “vectorized operation” mean in NumPy?
    -  In NumPy, a vectorized operation refers to applying a mathematical or logical operation to an entire array at once, completely eliminating the need to write explicit Python loops (like for or while).

        Instead of processing data element-by-element in slow, interpreted Python code, NumPy delegates the entire operation to highly optimized, pre-compiled C and Fortran code under the hood

15. How does NumPy handle multidimensional arrays?
    - NumPy handles multidimensional arrays by storing data in a contiguous, one-dimensional block of memory and using an indexing scheme (strides) to map multi-dimensional coordinates to that single block. This core object is known as the ndarray (N-dimensional array).By decoupling the physical memory from how the user interacts with the dimensions, NumPy achieves incredibly fast execution speeds and memory efficiency.

17.  Explain the difference between apply() and map() in Pandas.
 -  apply()

The apply() function is more flexible. It can be used with both Series and DataFrame objects.

For a Series, it can apply a function to each element. For a DataFrame, it can apply a function row-wise or column-wise using the axis parameter.

Example with Series:

marks = pd.Series([50, 60, 70, 80])

result = marks.apply(lambda x: x + 5)

print(result)

Example with DataFrame:

data = {
    'Maths': [70, 80, 90],
    'Science': [75, 85, 95]
}

df = pd.DataFrame(data)

result = df.apply(lambda x: x.mean())

print(result)

 - map()

The map() function is mainly used with a Pandas Series. It applies a function to each individual element of the Series and returns a new Series.

It is useful when we want to transform or modify each value separately.

Example:

import pandas as pd

marks = pd.Series([50, 60, 70, 80])

result = marks.map(lambda x: x + 5)

print(result)

```python
# How do you create a 2D NumPy array and calculate the sum of each row?
import numpy as np

# Creating a 2D NumPy array
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

# Calculating the sum of each row
row_sum = np.sum(arr, axis=1)

print("2D Array:")
print(arr)

print("Sum of each row:")
print(row_sum)
```

```python
# Write a Pandas script to find the mean of a specific column in a DataFrame?
import pandas as pd

# Creating a DataFrame
data = {
    "Name": ["Rahul", "Amit", "Priya", "Neha"],
    "Marks": [80, 75, 90, 85],
    "Age": [20, 21, 19, 20]
}

df = pd.DataFrame(data)

# Calculating the mean of Marks column
mean_marks = df["Marks"].mean()

print("DataFrame:")
print(df)

print("Mean of Marks:", mean_marks)
```

```python
# Create a scatter plot using Matplotlib
import matplotlib.pyplot as plt

# Data
hours = [1, 2, 3, 4, 5, 6]
marks = [40, 45, 55, 60, 70, 80]

# Creating scatter plot
plt.scatter(hours, marks)

# Adding labels and title
plt.xlabel("Hours Studied")
plt.ylabel("Marks Obtained")
plt.title("Relationship Between Study Hours and Marks")

# Display the graph
plt.show()
```

```python
#  How do you calculate the correlation matrix using Seaborn and visualize it with a heatmap
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Creating DataFrame
data = {
    "Math": [80, 85, 90, 70, 75],
    "Science": [75, 88, 92, 65, 78],
    "English": [70, 80, 85, 72, 76]
}

df = pd.DataFrame(data)

# Calculating correlation matrix
correlation = df.corr()

print("Correlation Matrix:")
print(correlation)

# Creating heatmap
sns.heatmap(correlation, annot=True, cmap="coolwarm")

plt.title("Correlation Matrix Heatmap")
plt.show()
```

```python
#  Create a line plot with multiple lines using Matplotlib.
import matplotlib.pyplot as plt

# X-axis values
months = ["Jan", "Feb", "Mar", "Apr", "May"]

# Data for two products
product_a = [100, 120, 140, 160, 180]
product_b = [80, 110, 130, 150, 170]

# Creating multiple lines
plt.plot(months, product_a, marker="o", label="Product A")
plt.plot(months, product_b, marker="o", label="Product B")

# Adding labels
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales Comparison")

# Displaying legend
plt.legend()

plt.show()
```

```python
#  Generate a Pandas DataFrame and filter rows where a column value is greater than a threshold.
import pandas as pd

# Creating DataFrame
data = {
    "Name": ["Rahul", "Amit", "Priya", "Neha", "Ravi"],
    "Marks": [80, 65, 90, 70, 85]
}

df = pd.DataFrame(data)

# Setting threshold
threshold = 75

# Filtering rows
filtered_df = df[df["Marks"] > threshold]

print("Original DataFrame:")
print(df)

print("\nStudents with marks greater than 75:")
print(filtered_df)
```

```python
#  Create a histogram using Seaborn to visualize a distribution.
import seaborn as sns
import matplotlib.pyplot as plt

# Student marks
marks = [45, 50, 55, 60, 62, 65, 68, 70,
         72, 75, 78, 80, 82, 85, 88, 90]

# Creating histogram
sns.histplot(marks, bins=5, kde=True)

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Student Marks")

plt.show()
```

```python
#  Perform matrix multiplication using NumPy.
import numpy as np

# Creating two matrices
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

# Matrix multiplication
result = np.matmul(A, B)

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

print("\nMatrix Multiplication:")
print(result)
```
